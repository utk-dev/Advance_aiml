```python
def entropy(p):
    if p == 0 or p == 1:
        return 0
    else:
        return -p*np.log2(p)-((1-p)*np.log2(1-p))
print(entropy(0.5))
```

```python
def split_indices(X, index_feature):
    left_indices = []
    right_indices = []
    for i,x in enumerate(X):
        if x[index_feature] == 1:
            left_indices.append(i)
        else:
            right_indices.append(i)
    return left_indices, right_indices
```
