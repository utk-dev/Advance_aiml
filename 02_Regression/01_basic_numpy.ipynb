```python
import numpy as np
import time

a = np.zeros(4); print(f"np.zeros:--> a = {a}, shape:--> a = {a.shape}")
b = np.zeros(4,); print(f"np.zeros:--> b = {b}, shape:--> b = {b.shape}")
c = np.random.random_sample(4); print(f"Random sample:--> c = {c}, shape of c:--> {c.shape} ")
```

```python
a = np.arange(4); print(a)
a = np.arange(4.); print(a)
a = np.random.rand(4); print(a)
```

```python
print("--------1D ARRAY--------")
a = np.arange(10)
print(a)
print(a[2].shape)
print("--------2D ARRAY--------")
b = np.array([[10,20,30],[40,50,60],[70,80,90]])
print(b)
print(b[2].shape)
print("----ELEMENT FROM END----")
print(a[-2])  #the counting from the end starts from '-1' and '0' unlike the couting from the beginning.
```

```python
print("----SLICING----")
print(a[1:10:3])
print("--PRINTING NEGATIVE VALUES--")  #this is called slicing--> syntax = [start:stop:step]
print(-a)
print("--SUM OF DIGITS--")
b = np.sum(a)
print(b)
print("----MEAN OF DIGITS----")
c = np.mean(a)
print(c)
print("----OPERTION----")
d = a**2
print(d)
```
