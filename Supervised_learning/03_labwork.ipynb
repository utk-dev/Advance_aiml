```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
```

```python
def compute_cost(x,y,w,b):
    m = x.shape[0] 

    cost_func = 0  #m=4 (0,1,2,3) if x = np.array([1,2,3,4])
    for i in range(m):
        f_wb = w*x[i] + b
        cost = (f_wb - y[i])**2
        cost_func += cost
    total_cost = (1/(2*m))*cost_func

    return total_cost
```

```python
w = 100
b = 200

cost = compute_cost(x_train, y_train, w, b)
print(cost)
```
